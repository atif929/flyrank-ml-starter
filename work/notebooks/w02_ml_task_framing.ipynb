## Target or Proxy

The target is whether a content page is showing a declining trend and should be considered for review. In the starter dataset, this is represented by the proxy label `trend_direction == "down"`. This is a defined rule based on observed search performance rather than a future outcome, so it serves as a proxy label for learning and experimentation.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

df["target"] = (df["trend_direction"] == "down").astype(int)

print(df[["trend_direction", "target"]].head())
print("\nPositive labels:", df["target"].sum())
print("Negative labels:", len(df) - df["target"].sum())
```
