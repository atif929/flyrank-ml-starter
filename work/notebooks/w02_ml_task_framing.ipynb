## Target or Proxy

The target is whether a content page is showing a declining trend and should be considered for review. In the starter dataset, this is represented by the proxy label `trend_direction == "down"`. This is a defined rule based on observed search performance rather than a future outcome, so it serves as a proxy label for learning and experimentation.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd

df = pd.read_csv("./flyrank-ml-starter/data/raw/content_refresh_anonymized.csv")

df["target"] = (df["trend_direction"] == "down").astype(int)

print(df[["trend_direction", "target"]].head())
print("\nPositive labels:", df["target"].sum())
print("Negative labels:", len(df) - df["target"].sum())
```

## Success Metric

The main success metric is **Precision@50**. This measures how many of the top 50 pages recommended by the model are actually good candidates for review. Since content teams usually review only a limited number of pages, Precision@50 is more useful than overall accuracy because it evaluates the quality of the highest-priority recommendations.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

success_metric = "Precision@50"

print("Success Metric:", success_metric)
print("Reason: Measures how accurate the model's top 50 recommendations are.")
```

Machine learning can learn these patterns from the data and rank pages more effectively than a single if-statement while still supporting transparent, evidence-based decisions.
